# Welcome to gopyter 🐹

A **Go notebook** in your terminal, building on
[GoNB](https://github.com/janpfeifer/gonb), the Go kernel for Jupyter.
Declarations persist across cells; statements run in `main()`.

```go
type Point struct{ X, Y float64 }

func (p Point) Dist() float64 {
	return math.Hypot(p.X, p.Y)
}
```

```go
p := Point{3, 4}
fmt.Println("distance:", p.Dist())
p
```

```go
func fib(n int) int {
	if n < 2 {
		return n
	}
	return fib(n-1) + fib(n-2)
}

for i := range 10 {
	fmt.Printf("fib(%d) = %d\n", i, fib(i))
}
```

```go
var sb strings.Builder
sb.WriteString("| n | n² |\n|---|---|\n")
for i := 1; i <= 4; i++ {
	fmt.Fprintf(&sb, "| %d | %d |\n", i, i*i)
}
nb.DisplayMarkdown(sb.String())
```

```go
for i := range 3 {
	fmt.Println("tick", i)
	time.Sleep(300 * time.Millisecond)
}
fmt.Fprintln(os.Stderr, "done (stderr)")
```
